```python
def agente(pergunta):
    info = busca_web(pergunta)
    prompt = f'''Você é um agente inteligente. O usuário fez a seguinte pergunta: "{pergunta}". Você encontrou estas informações na internet: {info}. Responda somente com base nas informações acima. Se não houver dados suficienes, diga que não foi possível responder com precisão'''
    
    resposta = client.chat.completions.create(
        model = 'gpt-3.5-turbo',
        messages=[{"role":"user", "content":prompt}]
    )
    
    return resposta.choices[0].message.content
```

```python
resposta = agente("Quem foi Akira Toryama?")
print(resposta)
```
