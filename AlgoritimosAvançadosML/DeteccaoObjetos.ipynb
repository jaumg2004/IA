```python
import os
import cv2
import numpy as np
```

```python
MODEL_CONFIG = r"C:\Users\Jaum\Desktop\CursoIA\5.Redes Neurais, Deep Learning e Computer Vision\MobileNetSSD_deploy.prototxt"
MODEL_WEIGHTS = r"C:\Users\Jaum\Desktop\CursoIA\5.Redes Neurais, Deep Learning e Computer Vision\MobileNetSSD_deploy.caffemodel"
CLASSES = ['background', 'aeroplane', 'bicycle', 'bird', 'boat', 'bottle', 'bus', 'car', 'chair', 'cat', 'cow', 'dinigtable', 'dog', 'horse', 'motorbike', 'person', 'pottedplant', 'sheep', 'sofa', 'train', 'tvmonitor']
```

```python
net = cv2.dnn.readNetFromCaffe(MODEL_CONFIG, MODEL_WEIGHTS)
```

```python
def detect_person_in_video(video_path):
    cap = cv2.VideoCapture(video_path)
    frame_count = 0
    while True:
        ret, frame = cap.read()
        if not ret:
            print('Fim do vídeo, ou erro de leitura')
            break
        frame_count += 1
        (h, w) = frame.shape[:2]
        blob = cv2.dnn.blobFromImage(frame, 0.007843, (300, 300), 127.5)
        net.setInput(blob=blob)
        detection = net.forward()
        
        for i in range(detection.shape[2]):
            confidence = detection[0, 0, i, 2]
            if confidence > 0.9:
                idx = int(detection[0, 0, i, 1])
                if CLASSES[idx] == 'person':
                    box = detection[0, 0, i, 3:7] * np.array([w, h, w, h])
                    (startX, startY, endX, endY) = box.astype(dtype=int)
                    
                    label = f'{CLASSES[idx]}: {confidence * 100:.2f}%'
                    cv2.rectangle(frame, (startX, startY), (endX, endY), (0, 255, 0), 2)
                    cv2.putText(frame, label, (startX, startY - 10), cv2.FONT_HERSHEY_SIMPLEX, 0.5, (0, 255, 0), 2)
                    print(f'Pessoa detectada no Frame {frame_count}')
                    cv2.imshow(frame)
                    
                    cap.release()
                    
                    return 'Pessoa detectada no vídeo'
        
        cap.release()
        return 'Nenhuma pessoa detectada no vídeo'
```

```python
detect_person_in_video(r'C:\Users\Jaum\Desktop\CursoIA\5.Redes Neurais, Deep Learning e Computer Vision\people.mp4')
```
