```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.svm import SVC
from sklearn.metrics import hamming_loss
from sklearn.multioutput import ClassifierChain
from sklearn.preprocessing import StandardScaler
```

```python
musica = pd.read_csv(r'C:\Users\Jaum\Desktop\CursoIA\4.Tópicos Avançados de Machine Learning\Musica.csv')
musica
```

```python
musica.shape
```

```python
classe = musica.iloc[:, 0:6].values
previsores = musica.iloc[:, 6:].values
```

```python
scaler = StandardScaler()
previsores = scaler.fit_transform(previsores)
previsores
```

```python
X_treinamento, X_teste, y_treinamento, y_teste = train_test_split(previsores, classe, test_size=0.3, random_state=5)
```

```python
svm = SVC()
clssifier_chain = ClassifierChain(base_estimator=svm, random_state=5)
clssifier_chain.fit(X_treinamento, y_treinamento)
clssifier_chain
```

```python
y_pred = clssifier_chain.predict(X_teste)
loss = hamming_loss(y_teste, y_pred)
print(f'Hamming loss: {loss}')
```
