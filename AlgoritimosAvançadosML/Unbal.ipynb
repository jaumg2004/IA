```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import LabelEncoder
from imblearn.over_sampling import SMOTENC
```

```python
credito = pd.read_csv(r'C:\Users\Jaum\Desktop\CursoIA\4.Tópicos Avançados de Machine Learning\credit_simple.csv', sep=';')
credito
```

```python
count = credito.groupby(['CLASSE']).size()
count
```

```python
y = credito['CLASSE'].values
X = credito.iloc[:, :-1].values
```

```python
labelencoder = LabelEncoder()
for i in range(X.shape[1]):
    if X[:, i].dtype == 'object':
        X[:, i] = labelencoder.fit_transform(X[:, i])
X
```

```python
sm = SMOTENC(random_state=0, categorical_features=[3, 5, 6])
X_res, y_res = sm.fit_resample(X, y)
```
