```python
import nltk
from nltk.corpus import stopwords
from nltk.stem import PorterStemmer, WordNetLemmatizer, SnowballStemmer, LancasterStemmer
from nltk.tokenize import word_tokenize, sent_tokenize
from nltk.tag import pos_tag, pos_tag_sents
import string
```

```python
nltk.download('stopwords')
nltk.download('punkt')
nltk.download('punkt_tab')
nltk.download('tagsets_json')
nltk.download('averaged_perceptron_tagger')
nltk.download('maxent_ne_chunker_tab')
nltk.download('wordnet')
nltk.download('words')
```

```python
Texto = '''Nada é mais miserável que o homem que peca secretamente e trata de tentar preservar a sua imagem'''
print(Texto)
```

```python
sentencas = sent_tokenize(Texto)
print(sentencas)
```

```python
print(len(sentencas))
```

```python
token = word_tokenize(Texto, language='portuguese')
print(token)
```

```python
print(len(token))
```

```python
stop = stopwords.words('portuguese')
print(stop)
```

```python
print(len(stop))
```

```python
palavras_sem_stopwords = [p for p in token if p not in stop]
print(token)
print(palavras_sem_stopwords)
```

```python
print(string.punctuation)
```

```python
palavras_sem_pontuacao = [p for p in palavras_sem_stopwords if p not in string.punctuation]
print(palavras_sem_stopwords)
print(palavras_sem_pontuacao)
print(token)
```

```python
frequencia = nltk.FreqDist(palavras_sem_pontuacao)
frequencia
```

```python
mais_comuns = frequencia.most_common(4)
mais_comuns
```

```python
stemmer = PorterStemmer()
stem1 = [stemmer.stem(word) for word in palavras_sem_stopwords]
print(palavras_sem_stopwords)
print(stem1)
```

```python
stemmer2 = SnowballStemmer('portuguese')
stem2 = [stemmer2.stem(word) for word in palavras_sem_stopwords]
print(palavras_sem_stopwords)
print(stem2)
```

```python
stemmer3 = LancasterStemmer()
stem3 = [stemmer3.stem(word) for word in palavras_sem_stopwords]
print(palavras_sem_stopwords)
print(stem3)
```

```python
nltk.help.upenn_tagset()
```

```python
pos = pos_tag(palavras_sem_pontuacao)
print(pos)
```

```python
lemmatizer = WordNetLemmatizer()
resultado = [lemmatizer.lemmatize(palavra) for palavra in palavras_sem_stopwords]
print(palavras_sem_stopwords)
print(resultado)
```

```python
texto_en = 'Barak Obama foi um presidente dos EUA'
token3 = word_tokenize(texto_en)
tags = pos_tag(token3)
en = nltk.ne_chunk(tags)
print(en)
```
